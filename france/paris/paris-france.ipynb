## Cour Napoléon - Pyramide du Louvre

The Palais du Louvre is a former royal palace located in the 1st arrondissement of Paris on the right bank of the Seine, between the Tuileries Gardens and the church of Saint-Germain-l'Auxerrois. 
Today, it houses the Louvre Museum. 
In the center of the Cour Napoléon, the Louvre pyramid seen from the first floor of the Colbert wing, next to the 3.15m high statue of Jean Racine, French dramatist of 17th-century.
